```python
import pandas as pd
archivo   = "data/heart.csv"  # <- su archivo
respuesta = "target"          # <- su respuesta
quitar    = []                # <- columnas a quitar

d = pd.read_csv(archivo).drop_duplicates()   # sin filas repetidas
if len(d) > 3000: d = d.sample(3000, random_state=2026)
y = d[respuesta]
X = d.drop(columns=[respuesta] + quitar)
for c in X.select_dtypes(exclude="number"):  # texto a números
    X[c] = X[c].astype("category").cat.codes
```

```python
from sklearn.model_selection import train_test_split
X_tr, X_te, y_tr, y_te = train_test_split(
    X, y, test_size=0.25, random_state=2026)
print(len(X_tr), len(X_te))      # filas de entrenamiento y de prueba
```

```python
import os
from google.colab import userdata

# Carga el token desde los secretos de Colab a las variables de entorno
os.environ["TABPFN_TOKEN"] = userdata.get("TABPFN_TOKEN")

# Opcional: desactiva el login interactivo por navegador
os.environ["TABPFN_NO_BROWSER"] = "1"
```

```python
from tabpfn import TabPFNClassifier

tab = TabPFNClassifier(random_state=1)
tab.fit(X_tr, y_tr)        # "lee" las filas de entrenamiento
pred = tab.predict(X_te)   # una clase para cada fila de prueba
pred[:10]
```

```python
error = (pred != y_te).mean()
base  = (y_te != y_tr.mode()[0]).mean()  # siempre la clase más común
print(f"Sin modelo: {base:.1%}   TabPFN: {error:.1%}")
```
